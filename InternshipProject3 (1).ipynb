## **Step by Step Problem**
**Step 1: Load & Explore Sales Data**

Before analyzing the sales data, we need to load it into Python and understand what kind of data we’re working with.

✅ Task:
Open the sales file and bring the data into Python

View a few starting records to get an idea of what’s inside

Check how many rows and columns the dataset has

See what types of data (text, number, date) are present in each column

```python
import pandas as pd
#to read the data of the csv file
data=pd.read_csv('datasets.csv')
#to get the record (10 rows)
print(data.head(10))
#to check the rows and columns
print("\n")
print(data.info())

print("\n")
print(f"No. of rows and columns are {data.shape} respectively")
```

**✅ Step 2: Clean and Preprocess the Data**

Before we can analyze the data, we must make sure it's clean and in the correct format.

✅ Task:
Check if any values are missing from the dataset

Make sure date columns are in the correct format

Confirm that each column has the right type of data (number, text, or date)

```python
import pandas as pd
#missing values in dataset
missing_info=data.isna()
print("Missing Value in each column: \n")
print(missing_info.sum())

#replacing values
#numeric values are replaced using .mean()
data['Amount']=data['Amount'].fillna(data['Amount'].mean())
data['Profit']=data['Profit'].fillna(data['Profit'].mean())
data['Quantity']=data['Quantity'].fillna(data['Quantity'].mean())

#non-numeric values are replaced using .mode()[0]
data['Category']=data['Category'].fillna(data['Category'].mode()[0])
data['Sub-Category']=data['Sub-Category'].fillna(data['Sub-Category'].mode()[0])
data['PaymentMode']=data['PaymentMode'].fillna(data['PaymentMode'].mode()[0])

#to convert order date column to datetime
data['Order Date']=pd.to_datetime(data['Order Date'])
print(data.head())

#to check and confirm the format
print(data.dtypes)
```

**✅ Plot 3: Top 10 States by Total Profit**

✅ Task:
We want to find which states are generating the most profit for the company.

To do this:

Group the data by State

Sum the Profit for each state

Sort the results in descending order

Plot the top 10 states using a horizontal bar chart

This helps the business decide which states to invest more in or review if low-performing.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate total profit per state
state_total_profit = data.groupby('State')['Profit'].sum()

# Sort states by total profit in descending order and get the top 10
top_10_states_data = state_total_profit.sort_values(ascending=False).head(10)

sns.set(style='whitegrid')
plt.figure(figsize=(12,6))

# Create a simple horizontal bar plot for the top 10 states by profit
sns.barplot(x=top_10_states_data.values, y=top_10_states_data.index,hue=top_10_states_data.index, palette='cividis')
plt.xlabel('Total Profit')
plt.ylabel('State')
plt.title('Top 10 States by Total Profit')
plt.tight_layout() # Adjust layout to prevent labels from being cut off
plt.show()
```

**✅ Step 4: Product Analysis**

* Products that are expensive but sold in small quantities (maybe overpriced or not promoted well)
* Categories that are most popular (so they know what’s in demand)

✅ Task:
In this step, we will use SQL to analyze product performance. You’ll help the company identify:

Group the data by sub-category, calculate average price and total quantity sold, and sort the results to find premium-priced products with low sales

Next, group the data by product category and sort by total quantity sold to identify most popular product categories

These insights help identify products that may need pricing adjustments or promotion, and categories that are high in demand.

```python
import pandas as pd
```

```python
# Group by 'Sub-Category' and calculate total 'Amount' (revenue) and total 'Quantity' sold
sub_category_summary = data.groupby('Sub-Category').agg(
    Total_Revenue=('Amount', 'sum'),
    Total_Quantity=('Quantity', 'sum')
).reset_index()

# Sort by Total Revenue in descending order and then by Total Quantity in ascending order
# This brings high-revenue, low-quantity products to the top.
expensive_low_volume_products = sub_category_summary.sort_values(
    by=['Total_Revenue', 'Total_Quantity'],
    ascending=[False, True]
)

print("\nSub-Categories with High Revenue but Low Quantity Sold (Top 10):")
display(expensive_low_volume_products.head(10))
```

```python
# Group by 'Category' and sum the 'Quantity' sold to find the most popular categories
popular_categories = data.groupby('Category')['Quantity'].sum().reset_index()

# Sort in descending order to see the most popular categories first
popular_categories = popular_categories.sort_values(by='Quantity', ascending=False)

print("\nMost Popular Product Categories by Quantity Sold:")
display(popular_categories)
```

**✅ Step 5: Data Visualization**

**📊 Plot 1: Top 10 Products by Revenue**

✅ Task:
We want to find which products generate the most money.

To do this:

Use groupby() on the Sub-Category column

Sum the Amount column to get total revenue per product type

Sort the results in descending order

Show the top 10 products using a vertical bar chart

This will help identify which products drive the most revenue.


```python
import matplotlib.pyplot as plt
import seaborn as sns

most_revenue_products = data.groupby('Sub-Category')['Amount'].sum().reset_index().sort_values(by='Amount', ascending=False).head(10)

sns.set(style='whitegrid')
plt.figure(figsize=(12,6))

# Create a simple horizontal bar plot for the top 10 products by revenue
sns.barplot(x='Amount', y='Sub-Category', data=most_revenue_products,hue='Sub-Category', palette='magma')
plt.xlabel('Total Revenue')
plt.ylabel('Sub-Category')
plt.title('Top 10 Products by Total Revenue')
plt.tight_layout() # Adjust layout to prevent labels from being cut off
plt.show()
```

**📉 Plot 2: Monthly Order Trend**

✅ Task:
We want to see how the number of orders changes month by month.

To do this:

Use the Year-Month column

Count how many orders happened each month using value_counts()

Sort the months so they appear in the correct order

Plot this as a line chart to visualize changes over time

This helps the business understand seasonal trends, growth, or any decline in performance.

```python
import seaborn as sns
import matplotlib.pyplot as plt

orders_count = data['Year-Month'].value_counts().sort_index(ascending=False).reset_index().head(10)
orders_count.columns = ['Year-Month', 'Number of Orders']

sns.set(style='whitegrid')
plt.figure(figsize=(12,6))

sns.lineplot(x='Year-Month',y='Number of Orders',data=orders_count,color='red')
plt.xlabel('Year-Month')
plt.ylabel('Number of Orders')
plt.title('Monthly Order Trend')
plt.tight_layout()
plt.show()
```

**✅ Plot 3: Top 10 States by Total Profit**

✅ Task:
We want to find which states are generating the most profit for the company.

To do this:

Group the data by State

Sum the Profit for each state

Sort the results in descending order

Plot the top 10 states using a horizontal bar chart

This helps the business decide which states to invest more in or review if low-performing.

```python
import seaborn as sns
import matplotlib.pyplot as plt

state_total_profit = data.groupby('State')['Profit'].sum().sort_values(ascending=False).reset_index().head(10)

sns.set(style='whitegrid')
plt.figure(figsize=(12,6))

sns.barplot(x='Profit',y='State',data=state_total_profit,hue='State',palette='viridis')
plt.xlabel('Profit')
plt.ylabel('State')
plt.title('Top 10 States by Total Profit')
plt.tight_layout()
plt.show()
```

**✅ Plot 4: Total Profit by Payment Mode**

✅ Task:
We want to understand which payment modes bring the highest total profit to the company.

To do this:

Group the dataset by PaymentMode

Calculate the total profit for each payment type

Sort the results in descending order

Plot the totals using a Seaborn bar plot

This will help the business focus on payment options that generate the most profit.

```python
import seaborn as sns
import matplotlib.pyplot as plt

highest_profit_payment_mode = data.groupby('PaymentMode')['Profit'].sum().sort_values(ascending=False).reset_index().head(10)

sns.set(style='whitegrid')
plt.figure(figsize=(12,6))

sns.barplot(x='Profit',y='PaymentMode',data=highest_profit_payment_mode, hue='PaymentMode', palette='rocket')
plt.xlabel('Profit')
plt.ylabel('Payment Mode')
plt.title('Total Profit by Payment Mode')
plt.tight_layout()
plt.show()
```

**✅ Step 7: Export Business Reports**

✅ Task:
In this final step, you will create and export a summary report that combines two different business insights:

**🔹 1. Top 10 Products by Revenue**
* Group the data by both Category and Sub-Category.
* Calculate the total revenue using the Amount column.
* Sort the results in descending order of revenue.
* Select the top 10 entries using .head(10).
* Reset the index to make the grouped values into columns.
* Rename the columns to clearly reflect:
["Top Product Category", "Sub-Category", "Total Revenue"].

**🔹 2. Monthly Order Volume**
* Use the Year-Month column to count the number of orders per month.
* Sort the months in chronological order.
* Convert the result into a DataFrame using .reset_index().
* Rename the columns as:
["Month", "Number of Orders"].

Take only the first 10 months to match the number of rows from the Top Products table.

**🔹 3. Combine and Export**
* Use pd.concat(..., axis=1) to join both tables side by side.
* Export the final combined DataFrame to a CSV file named: "top_products_and_monthly_orders.csv" using .to_csv().

```python
# 1. Top 10 Products by Revenue
revenue_products = data.groupby(['Category', 'Sub-Category'])['Amount'].sum().reset_index()
top_10_products_by_revenue = revenue_products.sort_values(by='Amount', ascending=False).head(10)
top_10_products_by_revenue.columns = ["Top Product Category", "Sub-Category", "Total Revenue"]

# 2. Monthly Order Volume
monthly_order_volume = data['Year-Month'].value_counts().sort_index().reset_index()
monthly_order_volume.columns = ["Month", "Number of Orders"]
monthly_order_volume = monthly_order_volume.head(10) # Take first 10 months to match row count

# 3. Combine and Export
final_report = pd.concat([top_10_products_by_revenue, monthly_order_volume], axis=1)

# Export to CSV
final_report.to_csv('top_products_and_monthly_orders.csv', index=False)

print("Report 'top_products_and_monthly_orders.csv' exported successfully!")
display(final_report)
```
